List

```python
a=[1, 2, 3,7,9,5,34] # ordered ,duplicate and mutable
a[0]=98
a.append(45)# adds value at end
a.insert(2,99)#insert at particular position
#a.remove(83)# removes the value
a.pop()#removes last element
a.sort(reverse=True)#sorts the list
a.sort()
a.extend([67,89,23])#adds multiple values at end
a.sort(reverse=True)
a.reverse()


b=[x for x in a if x%2==0]
try:
    print("the maximum is", max(a))
    print("the minimum is", min(a))
    print(a)
    print(b)
except Exception as e:
    print(e)
finally:
    print("result printed")
```
